```python
import yfinance as yf
import pandas as pd
```

```python
# define tickers and load data from the last 2 years
tickers = ["AAPL", "MSFT", "JPM"]
data = yf.download(tickers, period="2y")["Close"]

data.head()
```

```python
#temporarily saveing the data in a csv to access it from other notebooks
data.to_csv("../data/stock_data.csv")

print("Saved successfully!")
```

```python
#load metadata for defined tickers
meta = {}
for ticker in ["AAPL", "MSFT", "JPM"]:
    info = yf.Ticker(ticker).info
    meta[ticker] = {
        "sector": info.get("sector", "N/A"),
        "country": info.get("country", "N/A"),
        "industry": info.get("industry", "N/A")
    }
data.head()
```

```python
#saving metadata in separate CSV
meta_df = pd.DataFrame(meta).T
meta_df.to_csv("../data/metadata.csv")
print("Saved successfully!")
```
